```python
import torch
import dgl

# Keep the original graph unchanged
g_processed = g.clone()

# Convert features from float64 to float32
X = g.ndata["feature"].to(torch.float32).contiguous()

# Convert one-hot labels to class indices
y = g.ndata["label"].argmax(dim=1).long()

# Add processed data to our working graph
g_processed.ndata["feature"] = X
g_processed.ndata["class_label"] = y

# Verify preprocessing
print("PREPROCESSING SUMMARY")
print("-" * 45)

print("Graph nodes:", g_processed.num_nodes())
print("Graph edges:", g_processed.num_edges())

print("Feature shape:", X.shape)
print("Feature dtype:", X.dtype)

print("Label shape:", y.shape)
print("Label dtype:", y.dtype)

print("\nClass distribution:")
print("Class 0:", (y == 0).sum().item())
print("Class 1:", (y == 1).sum().item())

print("\nMissing values:", torch.isnan(X).sum().item())
print("Infinite values:", torch.isinf(X).sum().item())
```

DATA SPLIT


```python
import numpy as np
from sklearn.model_selection import train_test_split

# Convert labels to NumPy for splitting
y_np = y.numpy()

# All node indices
node_indices = np.arange(len(y_np))

# First: 10% training, 90% remaining
train_idx, remaining_idx = train_test_split(
    node_indices,
    test_size=0.90,
    random_state=27,
    stratify=y_np
)

# Second: split remaining 90% into 30% validation and 60% testing
val_idx, test_idx = train_test_split(
    remaining_idx,
    test_size=2/3,
    random_state=27,
    stratify=y_np[remaining_idx]
)

# Convert to PyTorch tensors
train_idx = torch.tensor(train_idx, dtype=torch.long)
val_idx = torch.tensor(val_idx, dtype=torch.long)
test_idx = torch.tensor(test_idx, dtype=torch.long)

# Verify split
print("DATA SPLIT SUMMARY")
print("-" * 40)

for name, indices in [
    ("Training", train_idx),
    ("Validation", val_idx),
    ("Testing", test_idx)
]:
    labels = y[indices]

    print(f"\n{name}:")
    print("Total nodes:", len(indices))
    print("Normal nodes:", (labels == 0).sum().item())
    print("Fraud nodes:", (labels == 1).sum().item())
```

```python
def create_labelled_training_set(
    train_idx,
    y,
    num_fraud_labels,
    seed=27
):
    rng = np.random.default_rng(seed)

    train_labels = y[train_idx]

    fraud_nodes = train_idx[train_labels == 1].numpy()
    normal_nodes = train_idx[train_labels == 0].numpy()

    # Sample the requested number of fraud nodes
    selected_fraud = rng.choice(
        fraud_nodes,
        size=num_fraud_labels,
        replace=False
    )

    # Preserve the original positive-to-negative ratio
    normal_count = round(
        num_fraud_labels * len(normal_nodes) / len(fraud_nodes)
    )

    selected_normal = rng.choice(
        normal_nodes,
        size=normal_count,
        replace=False
    )

    # Combine labelled nodes
    labelled_nodes = np.concatenate([
        selected_fraud,
        selected_normal
    ])

    # All remaining training nodes are unlabelled
    unlabelled_nodes = np.setdiff1d(
        train_idx.numpy(),
        labelled_nodes
    )

    return (
        torch.tensor(labelled_nodes, dtype=torch.long),
        torch.tensor(unlabelled_nodes, dtype=torch.long)
    )


# Experiment A: 1 labelled fraud node
labelled_1, unlabelled_1 = create_labelled_training_set(
    train_idx, y, num_fraud_labels=1
)

# Experiment B: 5 labelled fraud nodes
labelled_5, unlabelled_5 = create_labelled_training_set(
    train_idx, y, num_fraud_labels=5
)
```

```python
for name, labelled, unlabelled in [
    ("1 FRAUD LABEL", labelled_1, unlabelled_1),
    ("5 FRAUD LABELS", labelled_5, unlabelled_5)
]:
    print("\n", name)
    print("-" * 40)

    print("Labelled nodes:", len(labelled))
    print("Labelled fraud:", (y[labelled] == 1).sum().item())
    print("Labelled normal:", (y[labelled] == 0).sum().item())

    print("Unlabelled training nodes:", len(unlabelled))
```
